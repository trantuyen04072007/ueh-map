```
import folium
from folium import plugins
ueh_coords = [10.7828, 106.6958]
m = folium.Map(location=ueh_coords, zoom_start=16, control_scale=True)
fg_ueh = folium.FeatureGroup(name='Trường đại học')
fg_public = folium.FeatureGroup(name='Địa điểm công cộng')
folium.Marker(
    location=ueh_coords,
    popup="<b>UEH - Cơ sở A</b><br>Đại học Kinh tế TP.HCM",
    tooltip="Click để xem chi tiết",
    icon=folium.Icon(color='red', icon='university', prefix='fa')
).add_to(fg_ueh)
locations = [
    {"name": "Bệnh viện Nhi Đồng 2", "coords": [10.7835, 106.7010], "desc": "Bệnh viện chuyên khoa nhi tuyến đầu."},
    {"name": "Diamond Plaza", "coords": [10.7816, 106.6991], "desc": "Trung tâm thương mại và văn phòng."},
    {"name": "Nhà thờ Đức Bà", "coords": [10.7797, 106.6990], "desc": "Địa điểm du lịch và tôn giáo nổi tiếng."},
    {"name": "Bưu điện Trung tâm Thành phố", "coords": [10.7799, 106.7000], "desc": "Cơ quan hành chính và điểm tham quan lịch sử."},
    {"name": "Dinh Độc Lập", "coords": [10.7770, 106.6954], "desc": "Di tích lịch sử quốc gia đặc biệt."}
]
for loc in locations:
    folium.Marker(
        location=loc["coords"],
        popup=f"<b>{loc['name']}</b><br>{loc['desc']}",
        tooltip=loc["name"],
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(fg_public)
fg_ueh.add_to(m)
fg_public.add_to(m)
folium.LayerControl().add_to(m)
plugins.Fullscreen().add_to(m)
m.save('index.html')
m
```
